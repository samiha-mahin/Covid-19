# **Encode Yes/No → 1/0**

```python
import pandas as pd
import numpy as np

# Assuming your dataframe is called df
df_encoded = df.copy()

for col in df_encoded.columns:
    df_encoded[col] = df_encoded[col].map({'Yes': 1, 'No': 0})
```

# **Correlation Heatmap**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 10))
corr = df_encoded.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=True,
    fmt=".2f",
    annot_kws={"size": 8},
    linewidths=0.5
)

plt.title("Correlation Heatmap of COVID-19 Features")
plt.show()
```

# **Histogram (Distribution of Features)**

```python
df_encoded.hist(
    figsize=(30,15),
    bins=3,
    edgecolor="black"
)

plt.suptitle("Histogram of Binary Features", fontsize=16)
plt.show()
```

# **Bar Plot (COVID-19 Outcome Distribution)**

```python
covid_counts = df_encoded['COVID-19'].value_counts()

plt.figure(figsize=(6, 4))
covid_counts.plot(kind='bar', color="skyblue")

plt.xlabel("COVID-19 Status")
plt.ylabel("Count")
plt.title("COVID-19 Class Distribution")
plt.xticks(rotation=0)
plt.show()
```
